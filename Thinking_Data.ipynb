# Assignment - 1

```python
import json
```

```python
# Load Data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data, "\n",type(data))
```

```python
def clean_Data(data):
    text_to_num = {
        "one" : 1,
        "two" : 2,
        "three" : 3,
        "four" : 4,
        "five" : 5
    }
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # print(user) # Printing each list
        # Clean Rating data
        # Spaces after number, Float value, text
        # Extract Rating from user
        raw_rating = str(user["rating"]).strip().lower()
        # print(raw_rating)
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
            # The above line will replace the text to num
            # Eg: if we got four, text_to_num[four] -> 4
        user["rating"] = raw_rating
        # print(raw_rating)

        # Handle missing values
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # Deduplicate
        if(user["name"].strip() in unique_users):
            continue
        
        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
        
```

```python
data = clean_Data(data)
print(data[0])
print(data[1])
print(data[2])
print(data[3])
print(data[4])
```

# Assignment - 2
Get meaningful Insights

- What is average rating
- What % of user gave bad rating?

```python
# get meaningful Insights from data
def get_insights(data):

    #Get Average Rating!
    tot_rating = 0
    for user in data:
        tot_rating += float(user["rating"])

    print(f"Average rating = {tot_rating/len(data)}")

    # % of user with poor rating -> (poor rating / tot_rating) * 100
    poor_rating = 0
    for user in data:
        if (float(user["rating"]) < 3):
            poor_rating += 1
    
    print(f"Percentage of poor rating = {(poor_rating/len(data))*100}%")
        
```

```python
get_insights(data)
```

# Assignment - 3
Recommendation System (Rule based not ML)
- If user rating >= 4, Suggest Apple
- else suggest Samsung

```python
def get_recommendations(data):
    recommendation = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendation.append(curr_recomm)
    return recommendation
```

```python
get_recommendations(data)
```
