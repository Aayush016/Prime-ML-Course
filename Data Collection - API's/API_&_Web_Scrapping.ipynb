```python
from bs4 import BeautifulSoup

with open("scraped_data/data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'lxml')
```

```python
soup
```

```python
soup.find_all("h3")
```

```python
all_h3 = soup.find_all("h3")

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)
    # print(h3.find_parent("div")["class"])

    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    print(h3.find_next("div").select_one("span.country-population").get_text(strip=True))

    # print(h3.attrs)
    all_countries.append([name, population])
```

```python
all_countries
```

```python
import pandas as pd

df = pd.DataFrame(all_countries, columns=["Name", "Population"])

df
```

```python
df.to_csv("cleaned_data/data.csv", index=False)
```

# Web Scrapping Problems

## Problem - 1
From Quotes, extract all quotes & authors from all pages with life tag

```python
import requests
from bs4 import BeautifulSoup
```

```python
# Fetch the data and store it

# URL = "https://quotes.toscrape.com/page/1/"

"""
We face a problem here! Since the quotes are spread across multiple pages we need to visit them one by one?
How do we do that?
We can check for 2 end points. What are they?
When we get to an end page the web server will either send status code 404 or we will have an empty response.
For this particular quotes website we need to do the later one!

In the website we do not have any <div class="quote"> section when it comes to an empty result page
for page_count = 1 we will get a non empty list but if a empty quote page comes we will get empty list.
"""
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("No valid pages anymore...")
        break
        
    with open(f"scraped_data/quotes{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Page - {page_count} downloaded")

    page_count = page_count + 1
```

```python
# Extract useful information from downloaded pages.

with open("scraped_data/quotes1.html", "r", encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
all_quotes = soup.select("div.quote")
print(all_quotes)
```

```python
all_quotes = soup.select("div.quote")
life_quotes = []

for q in all_quotes:
    all_tags = []
    # .tags means the element and .tag is the class

    for tag in q.select(".tags .tag"):
        all_tags.append(tag.get_text())

    if "life" in all_tags:
        text = q.select_one("span.text").get_text()
        author = q.select_one("small.author").get_text()
        life_quotes.append([text, author])

print(life_quotes)
```
